1. 코랩 시작하기

구구단 만들기 코드를 for문을 써서 만들어줘 몇단을 할 지는 입력을 받아서 하게 해줘
for문을 사용하여 입력받은 단의 구구단을 출력하는 코드를 작성했습니다.



```python
dan = int(input('출력할 구구단의 단을 입력하세요 (예: 2): '))

print(f'\n--- {dan}단 ---')
for i in range(1, 10):
    print(f'{dan} x {i} = {dan * i}')
```

### 전체 구구단 (2단 ~ 9단) 출력하기
이번에는 중첩 for문을 사용하여 전체 구구단을 보기 좋게 출력하는 코드입니다.

```python
for dan in range(2, 10):
    print(f'\n--- {dan}단 ---')
    for i in range(1, 10):
        print(f'{dan} x {i} = {dan * i}')
```

### 큐피트의 화살과 하트 그리기
matplotlib를 활용하여 예쁜 하트와 하트를 관통하는 큐피트의 화살을 그려봅니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 하트 공식 데이터를 생성합니다 (parametric equation for heart shape)
t = np.linspace(0, 2 * np.pi, 1000)
x_heart = 16 * np.sin(t)**3
y_heart = 13 * np.cos(t) - 5 * np.cos(2*t) - 2 * np.cos(3*t) - np.cos(4*t)

# 그래프 크기 및 스타일 설정
plt.figure(figsize=(8, 8))
plt.fill(x_heart, y_heart, color='crimson', alpha=0.8, label='Heart')

# 큐피트의 화살 (시작점과 끝점을 연결하는 화살표)
# 화살은 왼쪽 위에서 오른쪽 아래로 비스듬히 하트를 관통하도록 설정합니다.
arrow_start_x = -22
arrow_start_y = 15
arrow_end_x = 22
arrow_end_y = -15

# 화살대 그리기
plt.plot([arrow_start_x, arrow_end_x], [arrow_start_y, arrow_end_y], color='gold', linewidth=4, label="Cupid's Arrow")

# 화살촉 그리기 (끝부분에 세모 모양)
plt.plot(arrow_end_x, arrow_end_y, marker='>', color='gold', markersize=15)

# 화살 깃 그리기 (시작 부분에 짧은 선들)
plt.plot([arrow_start_x, arrow_start_x + 3], [arrow_start_y, arrow_start_y - 1], color='cyan', linewidth=3)
plt.plot([arrow_start_x, arrow_start_x + 1], [arrow_start_y, arrow_start_y - 3], color='cyan', linewidth=3)

# 그래프 꾸미기
plt.title("Cupid's Arrow & Heart", fontsize=16, color='darkred', fontweight='bold')
plt.xlim(-25, 25)
plt.ylim(-25, 25)
plt.axis('off') # 격자와 축선 숨기기
plt.show()
```
